```python
from itertools import product

def generate_hospital_configurations():
    """
    Generate all possible hospital occupancy configurations.
    Each hospital can independently have occupancy rates from 90% to 95%.
    Total combinations = 6^6 = 46,656 configurations
    """
    hospitals = ["CHU-SJ", "CHUQ", "CHUS", "CUSM", "HGJ", "HMR"]
    occupancy_rates = [round(rate/100, 2) for rate in range(90, 96)]  # [0.90, 0.91, 0.92, 0.93, 0.94, 0.95]
    
    # Generate all possible combinations using itertools.product
    all_combinations = product(occupancy_rates, repeat=len(hospitals))
    
    # Convert combinations to configurations and print
    print("All possible hospital occupancy configurations:")
    print(f"Total configurations to generate: {len(occupancy_rates)**len(hospitals)}")
    print("-" * 80)
    
    configs = []
    for i, rates in enumerate(all_combinations, 1):
        config = {
            hospital: {"Intensive": rate, "Intermediate": rate}
            for hospital, rate in zip(hospitals, rates)
        }
        configs.append(config)
        
        # Print progress every 1000 configurations
        if i % 1000 == 0:
            print(f"Generated {i} configurations...")
        
        # Print first and last few configurations as examples
        if i <= 3 or i > (6**6 - 3):
            print(f"\nConfiguration {i}:")
            print("{")
            for hospital, values in config.items():
                print(f'    "{hospital}": {values},')
            print("}")
            print("-" * 80)
    
    print(f"\nTotal configurations generated: {len(configs)}")
    return configs

# Run the function
configurations = generate_hospital_configurations()
```
